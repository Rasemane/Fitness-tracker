```python
#47933062_Rasemane
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
#47933062_Rasemane
# Randomly create the number of users for each cluster
np.random.seed(42)
user_cluster1 = np.random.randint(100, 1000)
user_cluster2 = np.random.randint(100, 1000)
user_cluster3 = np.random.randint(100, 1000)

# Generate data for each cluster
steps_cluster1 = np.random.randint(8000, 12000, user_cluster1)
sleep_cluster1 = np.random.uniform(7, 9, user_cluster1)

steps_cluster2 = np.random.randint(5000, 8000, user_cluster2)
sleep_cluster2 = np.random.uniform(6, 7.5, user_cluster2)

steps_cluster3 = np.random.randint(2000, 5000, user_cluster3)
sleep_cluster3 = np.random.uniform(5, 6.5, user_cluster3)

# Combine data into a single dataset
steps = np.concatenate([steps_cluster1, steps_cluster2, steps_cluster3])
sleep = np.concatenate([sleep_cluster1, sleep_cluster2, sleep_cluster3])

data = np.column_stack((steps, sleep))
```

```python
#47933062_Rasemane
# Scatter plot of the generated data
plt.figure(figsize=(8, 6))
plt.scatter(steps, sleep, alpha=0.6)
plt.title("Fitness Tracker Data Visualization")
plt.xlabel("Steps per Day")
plt.ylabel("Hours Slept per Night")
plt.show()
```

```python
#47933062_Rasemane
# Apply KMeans clustering
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(data)

# Retrieve cluster labels
labels = kmeans.labels_
```

```python
#47933062_Rasemane
# Visualize the clustered data
plt.figure(figsize=(8, 6))
for cluster in range(3):
    cluster_data = data[labels == cluster]
    plt.scatter(cluster_data[:, 0], cluster_data[:, 1], label=f"Cluster {cluster}")

plt.title("KMeans Clustering of Fitness Tracker Data")
plt.xlabel("Steps per Day")
plt.ylabel("Hours Slept per Night")
plt.legend()
plt.show()
```

```python
#47933062_Rasemane
# Calculate average steps and sleep values for each cluster
for cluster in range(3):
    cluster_data = data[labels == cluster]
    avg_steps = np.mean(cluster_data[:, 0])
    avg_sleep = np.mean(cluster_data[:, 1])
    print(f"Cluster {cluster}: Average Steps = {avg_steps:.2f}, Average Sleep = {avg_sleep:.2f} hours")
```
